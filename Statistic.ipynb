Outliers

```python
import numpy as np
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,27,34,37,201])
arr
```

```python
q1 = np.percentile(arr,25)
q2 = np.percentile(arr,50)
q3 = np.percentile(arr,75)
print(q1,q2,q3)
```

```python
iqr = q3 - q1
iqr
```

```python
uf = q3 + 1.5* iqr
uf
```

```python
lf = q3 - 1.5* iqr
lf
```

```python
l = []
for i in arr:
  if i >= lf and i <= uf:
    l.append(i)
l
```

```python
print(arr)
print(l)
```

```python
import seaborn as sns
sns.boxplot(x = l)
```
