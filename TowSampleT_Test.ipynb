```python
import numpy as np
from scipy import stats
```

```python
gp01 = [85,88,90,92,87,85,89,91,86,88]
gp02 = [82,84,80,83,81,79,78,85,84,83]
```

```python
t_stats, p_vaslue = stats.ttest_ind(gp01, gp02, equal_var = False)
# print(t_stats, p_vaslue)
```

```python
t_stats
```

```python
alpha = 0.5
```

```python
if p_vaslue < alpha:
  print(" i will reject null hypothesis")
else:
  print(" i will accept null hypothesis")
```
