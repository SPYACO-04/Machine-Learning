Pivot Tables

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values = "Sales",index = 'Region', columns='Product')
```

```python
pd.pivot_table(df,values = ["Sales","Units"],index = 'Region', columns='Product')
```

Cross Tables

```python
pd.crosstab(df['Region'],df['Product'])
```

```python
# Loading the data
df = pd.read_csv("anime.csv")
```

```python
def extract_ep(txt):
  check = False
  data = ""
  for i in txt :
    if i == ")":
      check = False
      return data
    if check == True :
      data = data + i
    if i == "(" :
      check = True

```

```python

df["Title"].apply(extract_ep)
```

```python
df["Episods"]= df["Title"].apply(extract_ep)
```

```python
df
```

```python
df["Episods"] = df["Episods"].str.replace("eps","")
```

```python
df["Episods"]=df["Episods"].astype(int)
```

```python
df
```

```python
df.loc[0]["Title"]
```

```python
def extractiom_time(txt):
  check = False
  data = ""
  for i in range(len(txt)) :
    if txt[i] == ')' :
      for j in range(i+1, i+20):
        data += txt[j]
      return data
```

```python
df["Total Time"] = df["Title"].apply(extractiom_time)
```
